```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq
from scipy.signal import welch

# Set global plotting parameters
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 10

# =============================================================================
# QUESTION 1: Basic Sinusoidal Signal
# =============================================================================

def question_1():
    print("=" * 60)
    print("QUESTION 1: Basic Sinusoidal Signal")
    print("=" * 60)

    # (a) Generate a basic sinusoidal signal
    print("\n(a) Generating sinusoidal signal...")
    fs = 1000  # Sampling frequency (1000 Hz)
    T = 5.0    # Duration in seconds
    f = 10      # Frequency (5 Hz)

    # Time vector
    t = np.linspace(0, T, int(fs * T), endpoint=False)

    # Generate sine wave
    signal = np.sin(2 * np.pi * f * t)

    # (b) Plot time-domain waveform
    print("(b) Plotting time-domain waveform...")
    plt.figure(figsize=(15, 10))

    plt.subplot(2, 2, 1)
    plt.plot(t, signal)
    plt.title('Question 1(b): Time-Domain Waveform\n5 Hz Sine Wave')
    plt.xlabel('Time (seconds)')
    plt.ylabel('Amplitude')
    plt.grid(True)
    plt.xlim(0, 0.5)  # Show first 0.5 seconds for clarity

    # (c) Compute and plot DTFT (using Welch's method for estimation)
    print("(c) Computing DTFT...")
    frequencies_dtft, spectrum_dtft = welch(signal, fs, nperseg=1024)

    plt.subplot(2, 2, 2)
    plt.plot(frequencies_dtft, spectrum_dtft)
    plt.title('Question 1(c): DTFT - Continuous Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 50)  # Focus on relevant frequency range

    # (d) Compute and plot DFT
    print("(d) Computing DFT...")
    N = len(signal)
    dft = fft(signal)
    frequencies_dft = fftfreq(N, 1/fs)

    # Take only positive frequencies
    positive_freq_idx = frequencies_dft >= 0
    frequencies_dft_positive = frequencies_dft[positive_freq_idx]
    dft_positive = np.abs(dft[positive_freq_idx])

    plt.subplot(2, 2, 3)
    plt.stem(frequencies_dft_positive, dft_positive, basefmt=" ")
    plt.title('Question 1(d): DFT - Discrete Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 50)

    plt.tight_layout()
    plt.show()

    return t, signal, frequencies_dtft, spectrum_dtft, frequencies_dft_positive, dft_positive

# =============================================================================
# MAIN EXECUTION
# =============================================================================

if __name__ == "__main__":
    print("Speech Processing Lab Exercise")
    print("Digital Signal Processing - Time and Frequency Domain Analysis")
    print("=" * 60)

    # Execute all questions
    q1_results = question_1()
```

```python
# =============================================================================
# QUESTION 2: Composite Signal
# =============================================================================

def question_2():
    print("\n" + "=" * 60)
    print("QUESTION 2: Composite Signal")
    print("=" * 60)

    # (a) Generate composite signal
    print("\n(a) Generating composite signal...")
    fs = 1000  # Sampling frequency
    T = 1.0    # Duration in seconds

    t = np.linspace(0, T, int(fs * T), endpoint=False)

    # Create three sinusoidal components with different frequencies and amplitudes
    f1, A1 = 10, 1.0   # 10 Hz with amplitude 1
    f2, A2 = 25, 0.5   # 25 Hz with amplitude 0.5
    f3, A3 = 50, 0.3   # 50 Hz with amplitude 0.3

    signal1 = A1 * np.sin(2 * np.pi * f1 * t)
    signal2 = A2 * np.sin(2 * np.pi * f2 * t)
    signal3 = A3 * np.sin(2 * np.pi * f3 * t)

    composite_signal = signal1 + signal2 + signal3

    # (b) Plot time-domain waveform
    print("(b) Plotting time-domain waveform...")
    plt.figure(figsize=(15, 10))

    plt.subplot(2, 2, 1)
    plt.plot(t, composite_signal)
    plt.title('Question 2(b): Composite Signal Time-Domain Waveform')
    plt.xlabel('Time (seconds)')
    plt.ylabel('Amplitude')
    plt.grid(True)
    plt.xlim(0, 0.3)  # Show first 0.3 seconds

    # (c) Compute and plot DTFT
    print("(c) Computing DTFT...")
    frequencies_dtft, spectrum_dtft = welch(composite_signal, fs, nperseg=1024)

    plt.subplot(2, 2, 2)
    plt.plot(frequencies_dtft, spectrum_dtft)
    plt.title('Question 2(c): DTFT - Continuous Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 100)

    # (d) Compute and plot DFT
    print("(d) Computing DFT...")
    N = len(composite_signal)
    dft = fft(composite_signal)
    frequencies_dft = fftfreq(N, 1/fs)

    positive_freq_idx = frequencies_dft >= 0
    frequencies_dft_positive = frequencies_dft[positive_freq_idx]
    dft_positive = np.abs(dft[positive_freq_idx])

    plt.subplot(2, 2, 3)
    plt.stem(frequencies_dft_positive, dft_positive, basefmt=" ")
    plt.title('Question 2(d): DFT - Discrete Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 100)

    # Mark the frequency components
    plt.axvline(x=f1, color='red', linestyle='--', alpha=0.7, label=f'{f1} Hz')
    plt.axvline(x=f2, color='green', linestyle='--', alpha=0.7, label=f'{f2} Hz')
    plt.axvline(x=f3, color='blue', linestyle='--', alpha=0.7, label=f'{f3} Hz')
    plt.legend()

    plt.tight_layout()
    plt.show()

    return t, composite_signal, frequencies_dtft, spectrum_dtft, frequencies_dft_positive, dft_positive

# =============================================================================
# MAIN EXECUTION
# =============================================================================

if __name__ == "__main__":
    print("Speech Processing Lab Exercise")
    print("Digital Signal Processing - Time and Frequency Domain Analysis")
    print("=" * 60)

    # Execute all questions
    q1_results = question_2()
```

```python
# =============================================================================
# QUESTION 3: Exponentially Decaying Signal
# =============================================================================

def question_3():
    print("\n" + "=" * 60)
    print("QUESTION 3: Exponentially Decaying Signal")
    print("=" * 60)

    # (a) Generate exponentially decaying signal
    print("\n(a) Generating exponentially decaying signal...")
    fs = 1000  # Sampling frequency
    T = 2.0    # Duration in seconds (longer to show decay)

    t = np.linspace(0, T, int(fs * T), endpoint=False)

    # Parameters for the decaying signal
    f_carrier = 20  # Carrier frequency (Hz)
    decay_rate = 3   # Decay rate

    # Create exponentially decaying sine wave
    signal = np.exp(-decay_rate * t) * np.sin(2 * np.pi * f_carrier * t)

    # (b) Plot time-domain waveform
    print("(b) Plotting time-domain waveform...")
    plt.figure(figsize=(15, 12))

    plt.subplot(2, 2, 1)
    plt.plot(t, signal)
    plt.title('Question 3(b): Exponentially Decaying Signal\nTime-Domain Waveform')
    plt.xlabel('Time (seconds)')
    plt.ylabel('Amplitude')
    plt.grid(True)

    # (c) Compute and plot DTFT
    print("(c) Computing DTFT...")
    frequencies_dtft, spectrum_dtft = welch(signal, fs, nperseg=1024)

    plt.subplot(2, 2, 2)
    plt.plot(frequencies_dtft, spectrum_dtft)
    plt.title('Question 3(c): DTFT - Continuous Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 100)

    # (d) Compute and plot DFT
    print("(d) Computing DFT...")
    N = len(signal)
    dft = fft(signal)
    frequencies_dft = fftfreq(N, 1/fs)

    positive_freq_idx = frequencies_dft >= 0
    frequencies_dft_positive = frequencies_dft[positive_freq_idx]
    dft_positive = np.abs(dft[positive_freq_idx])

    plt.subplot(2, 2, 3)
    plt.stem(frequencies_dft_positive, dft_positive, basefmt=" ")
    plt.title('Question 3(d): DFT - Discrete Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 100)

    plt.tight_layout()
    plt.show()

    # (e) Analyze relationship between time-domain and frequency-domain
    print("\n(e) Analysis:")
    print("• The exponentially decaying signal shows a dominant frequency at 20 Hz")
    print("• The decay in time domain results in spectral broadening around the carrier frequency")
    print("• Unlike pure sine waves, decaying signals have wider frequency spectra")
    print("• The faster the decay, the wider the frequency spectrum becomes")
    print("• This demonstrates the time-frequency uncertainty principle")

    return t, signal, frequencies_dtft, spectrum_dtft, frequencies_dft_positive, dft_positive

# =============================================================================
# MAIN EXECUTION
# =============================================================================

if __name__ == "__main__":
    print("Speech Processing Lab Exercise")
    print("Digital Signal Processing - Time and Frequency Domain Analysis")
    print("=" * 60)

    # Execute all questions
    q1_results = question_3()
```

```python
# =============================================================================
# QUESTION 4: Rectangular Pulse Signal
# =============================================================================

def question_4():
    print("\n" + "=" * 60)
    print("QUESTION 4: Rectangular Pulse Signal")
    print("=" * 60)

    # (a) Generate rectangular pulse signal
    print("\n(a) Generating rectangular pulse signal...")
    fs = 1000  # Sampling frequency
    T = 2.0    # Total duration in seconds

    t = np.linspace(0, T, int(fs * T), endpoint=False)

    # Create rectangular pulse (1 for 0.2 seconds, 0 elsewhere)
    pulse_duration = 0.2  # seconds
    pulse_start = 0.5     # seconds

    signal = np.zeros_like(t)
    pulse_mask = (t >= pulse_start) & (t < pulse_start + pulse_duration)
    signal[pulse_mask] = 1.0

    # (b) Plot time-domain waveform
    print("(b) Plotting time-domain waveform...")
    plt.figure(figsize=(15, 12))

    plt.subplot(2, 2, 1)
    plt.plot(t, signal)
    plt.title('Question 4(b): Rectangular Pulse\nTime-Domain Waveform')
    plt.xlabel('Time (seconds)')
    plt.ylabel('Amplitude')
    plt.grid(True)
    plt.ylim(-0.1, 1.1)

    # (c) Compute and plot DTFT
    print("(c) Computing DTFT...")
    frequencies_dtft, spectrum_dtft = welch(signal, fs, nperseg=1024)

    plt.subplot(2, 2, 2)
    plt.plot(frequencies_dtft, spectrum_dtft)
    plt.title('Question 4(c): DTFT - Continuous Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 100)

    # (d) Compute and plot DFT
    print("(d) Computing DFT...")
    N = len(signal)
    dft = fft(signal)
    frequencies_dft = fftfreq(N, 1/fs)

    positive_freq_idx = frequencies_dft >= 0
    frequencies_dft_positive = frequencies_dft[positive_freq_idx]
    dft_positive = np.abs(dft[positive_freq_idx])

    plt.subplot(2, 2, 3)
    plt.stem(frequencies_dft_positive, dft_positive, basefmt=" ")
    plt.title('Question 4(d): DFT - Discrete Frequency Spectrum')
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude')
    plt.grid(True)
    plt.xlim(0, 100)

    plt.tight_layout()
    plt.show()

    # (e) Analyze relationship between time-domain and frequency-domain
    print("\n(e) Analysis:")
    print("• Rectangular pulse in time domain transforms to sinc function in frequency domain")
    print("• The narrower the pulse in time, the wider the main lobe in frequency")
    print("• The spectrum shows characteristic sinc pattern with side lobes")
    print("• This demonstrates the inverse relationship between time and frequency resolution")
    print("• Zero crossings occur at multiples of 1/pulse_duration = 5 Hz")

    return t, signal, frequencies_dtft, spectrum_dtft, frequencies_dft_positive, dft_positive

# =============================================================================
# MAIN EXECUTION
# =============================================================================

if __name__ == "__main__":
    print("Speech Processing Lab Exercise")
    print("Digital Signal Processing - Time and Frequency Domain Analysis")
    print("=" * 60)

    # Execute all questions
    q1_results = question_4()
```
